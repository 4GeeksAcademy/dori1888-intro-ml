# 01 — Exploración y limpieza


Objetivo: entender el dataset, revisar `income`, limpiar valores mal codificados y guardar un CSV limpio.

**Salida:** `data/adult_clean.csv`


## 1) Carga de datos


```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/adult-census-income.csv")

df.head()
```

```python
df.shape, 
df.columns
```

```python
df.info()
```

## 2) Variable objetivo (income)


```python
df["income"].value_counts()
```

```python
df["income"].value_counts(normalize=True)
```

```python
(df == "?").sum().sort_values(ascending=False)
```

## 3) Limpieza


```python
df = df.replace("?", np.nan)
df = df.dropna()
df.shape
```

```python
num_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(include="object").columns.drop("income")

num_cols, cat_cols
```

## 4) Guardado del dataset limpio


```python
df.to_csv("../data/adult_clean.csv", index=False)
```
